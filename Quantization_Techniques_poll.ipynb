### Uniform Quantization
Based on [Improving the speed of neural networks on CPUs (2011)](https://research.google.com/pubs/pub37631.html?source=post_page)

```python
import numpy as np

"""
Quantize floating-point tensor X to b-bit integer representation.
Returns:
    X_q : quantized tensor
    s   : scale
    z   : zero-point
"""
def quantize(X, b):
    X = np.asarray(X, dtype=np.float32)
    x_min = min(np.min(X), 0.0)
    x_max = max(np.max(X), 0.0)
    X = np.clip(X, x_min, x_max)
    q_min = 0
    q_max = 2**b - 1
    s = (x_max - x_min) / (q_max - q_min)
    if s == 0:
        return np.zeros_like(X, dtype=np.int32), 1.0, 0
    z = round(q_min - x_min / s)
    X_q = np.round(X / s) + z
    X_q = np.clip(X_q, q_min, q_max)
    return X_q.astype(np.int32), s, z


"""
Dequantize integer tensor X_q back to floating point.
"""
def dequantize(X_q, s, z):
    X_hat = s * (X_q.astype(np.float32) - z)

    return X_hat
```
